```python
import pandas as pd
```

```python
df=pd.DataFrame({
    "Name":["pavan","harpal","rahul","bhau"],
    "Date":["17-06-2005","15-04-2004","05-12-2026","11-05-2000"]
})
df
```

```python
df["Date"]=pd.to_datetime(
    df["Date"],
    dayfirst=True,
    errors="coerce"
)
```

```python
df
```

```python
df["Day"]=df["Date"].dt.day
df["Month"]=df["Date"].dt.month
df["Year"]=df["Date"].dt.year
```

```python
df
```

```python
# merge()
```

```python
student=pd.DataFrame(
    {
        "S_id":[101,102,103],
        "Name":["pavan","harpal","bhau"]
    }
)

marks=pd.DataFrame(
    {
        "S_id":[101,102,103],
        "Marks":[85,95,34]
    }
)
result=pd.merge(student,marks,on="S_id")
result
```

```python
# join()
```

```python
student=pd.DataFrame(
    {
        "Name":["pavan","harpal","bhau"]
    },
    index=[101,102,103]
)

marks=pd.DataFrame(
    {
        "Marks":[85,95,34]
    },
    index=[101,102,103],
)
result=student.join(marks)
result
```

```python
# concat
```

```python
class_a = pd.DataFrame(
    {
        "Name":["pavan","patel"],
        "Marks":[54,94]
    }
)

class_b = pd.DataFrame(
    {
        "Name":["harpal","bhau"],
        "Marks":[84,42]
    }
)
result = pd.concat(
    [class_a,class_b],
    ignore_index=True
)
result
```
