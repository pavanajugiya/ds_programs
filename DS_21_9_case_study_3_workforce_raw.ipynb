```python
import pandas as pd
import numpy as np

work = pd.read_csv("case_study_3_workforce_raw.csv")
work
```

```python
dept = pd.read_csv("case_study_3_department_master.csv")
dept
```

```python
work_clean=work.copy()
work_clean.loc[work_clean.duplicated()]
```

```python
work_clean=work_clean.drop_duplicates()
```

```python
work_clean.loc[work_clean.duplicated()]
```

```python
work_clean.loc[work_clean.isnull().any(axis=1)]
```

```python
work_clean["Experience_Years"]=work_clean["Experience_Years"].fillna(work_clean.groupby("Department")["Experience_Years"].transform("mean"))
work_clean["Monthly_Salary"]=work_clean["Monthly_Salary"].fillna(work_clean.groupby("Department")["Monthly_Salary"].transform("mean"))
work_clean["Performance_Rating"]=work_clean["Performance_Rating"].fillna(work_clean.groupby("Department")["Performance_Rating"].transform("mean"))
```

```python
work_clean.loc[work_clean.isnull().any(axis=1)]
```

```python
work_clean
```

```python
work_clean.sort_values(by=["Monthly_Salary","Employee_Name","Department"],ascending=False).head()
```

```python
av=work_clean.groupby("Department")[["Performance_Rating","Monthly_Salary"]].mean()
```

```python
av.sort_values(by="Monthly_Salary",ascending=False).head(1)
```

```python
work_clean["Employee_Name"] = work_clean["Employee_Name"].str.title().str.strip()
work_clean["Employee_Number"] = (work_clean["Employee_Code"].str.extract(r"(\d+)$")[0].astype(int))
work_clean[["Employee_Code", "Employee_Name", "Employee_Number"]]
```

```python
department_distribution = (work_clean["Department"].value_counts().reset_index())
print(department_distribution)
department_distribution.columns = ["Department", "Employee_Count"]
department_distribution["Percentage"] = (department_distribution["Employee_Count"]/ len(work_clean) * 100).round(2)
print(department_distribution)
```

```python
gender_department = pd.crosstab(work_clean["Department"],work_clean["Gender"])
gender_department
```

```python
work_clean["Join_Date"] = pd.to_datetime(
    work_clean["Join_Date"],
    errors="coerce"
)

work_clean["Join_Year"] = work_clean["Join_Date"].dt.year

joining_by_year = (
    work_clean["Join_Year"]
    .value_counts()
    .sort_index()
)

print(joining_by_year)
```

```python
most_joining_year = joining_by_year.idxmax()
maximum_joining = joining_by_year.max()

print(most_joining_year)
print(maximum_joining)

merged_df = pd.merge(work_clean,dept,on="Department",how="left")

department_master_summary = merged_df.groupby(["Department", "Business_Unit", "Department_Head"]).agg(
    Employee_Count=("Employee_Code", "count"),
    Average_Salary=("Monthly_Salary", "mean")
).round(2)

display(department_master_summary)
```
