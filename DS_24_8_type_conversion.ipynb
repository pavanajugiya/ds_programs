```python
df1=df.copy()
df1["StudentID"]=df1["StudentID"].astype("string") #convert into string
df1["StudentID"].dtype
```

```python
df1["Fees"]=df1["Fees"].astype(float) #convert into float
df1["Fees"].dtype
```

```python
df1.dtypes
```

```python
df1["City"]=df1["City"].astype("category") #convert into category
df1["City"].dtype
```

```python
df1["City"].cat.categories # shows all category
```

```python
df1["Age"] = pd.to_numeric(df1["Age"],errors="coerce").astype("Int64") # convet into int and remove wrong values = NaN 
df1["Age"]
```

```python
age_numeric = pd.to_numeric(df["Age"],errors="raise") # exit when error comes
```

```python
age_numeric1 = pd.to_numeric(df["Age"],errors="ignore") # ignore not convertable values
```

```python
df1["JoinDate"] = pd.to_datetime(
    df1["JoinDate"],
    format="mixed",
    errors="coerce") # convet into date formate
df1["JoinDate"]
```

```python
df1["StudyTime"] = pd.to_timedelta(
    df1["StudyTime"],
    errors="coerce") # convert into timedelta(day & hours) type
df1["StudyTime"]
```

```python
active_map = {
    "yes":True,"y":True,"true":True,"1":True,
    "no":False,"n":False,"false":False,"0":False
}
df1["Active"] = (
    df1["Active"]
    .astype("string")
    .str.strip()
    .str.lower()
    .fillna("False")
    .map(active_map)
    .astype("boolean")
) # convert into boolean
```
