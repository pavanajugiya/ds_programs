```python
import pandas as pd
```

```python
df=pd.read_csv("students_type_conversion.csv")
df
```

```python
df1 = df.copy()                                                                                                                                         
```

```python
active_map = {
    "yes":True,"y":True,"true":True,"1":True,
    "no":False,"n":False,"false":False,"0":False
}
df1["Active"] = (
    df["Active"]
    .astype("string")
    .str.strip()
    .str.lower()
    .map(active_map)
    .astype("boolean")
)
```
