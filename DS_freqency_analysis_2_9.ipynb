```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Name": ["Aarav", "Diya", "Rohan", "Priya", "Kabir", "Meera"],
    "Course": ["MCA", "MBA", "MCA", "MCA", "MBA", "MCA"],
    "Gender": ["M", "F", "M", "F", "M", "F"],
    "Marks": [82, 91, 76, 88, 69, 95],
    "Study_Hours": [4, 6, 3, 5, 2, 7]
})
df
```

```python
print(df["Course"].value_counts())
```

```python
print(df[["Course","Gender"]].value_counts())
```

```python
df["Course"].value_counts(normalize=True)
```

```python
df["Course"].value_counts(normalize=True)*100
```

```python
round(df["Course"].value_counts(normalize=True)*100)
```

```python
df[["Course","Gender"]].value_counts(normalize=True)*100
```

```python
df["Course"].value_counts(normalize=False)
```

```python
pd.crosstab(df["Course"],df["Gender"])
```

```python
pd.crosstab(df["Course"],df["Gender"],margins=True)
```

```python
pd.pivot_table(
    df,
    values="Marks",
    index="Course",
    columns="Gender",
    aggfunc="mean"
)
```

```python
pd.pivot_table(
    df,
    values="Marks",
    index="Gender",
    columns="Course",
    aggfunc="mean"
)
```

```python
df.pivot_table(
    values="Marks",
    index="Course",
    aggfunc=["mean","max"]
)
```

```python
df.pivot_table(
    values="Marks",
    index="Course",
    columns="Gender",
    aggfunc=["mean","max"]
)
```
