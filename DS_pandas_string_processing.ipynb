```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Name": ["  Aarav Shah  ", "diya PATEL", "Rohan Mehta", "priya joshi"],
    "Email": ["AARAV@GMAIL.COM", "diya@yahoo.com", "rohan@gmail.com", "priya@outlook.com"],
    "Course_Code": ["MCA-101", "BCA-205", "MCA-310", "BSC-115"]
})
```

```python
df
```

```python
df["Email"] = df["Email"].str.lower()  #convert into lowercase
```

```python
df["Course_Code"] = df["Course_Code"].str.upper() # convert into uppercase
```

```python
df["Name"] = df["Name"].astype("string").str.strip() # remove whitespaces 
```

```python
df
```

```python
price = pd.Series(["₹1,200", "₹850", "₹2,500"])
price
```

```python
price=(price.astype("string")
       .str.replace("₹","",regex=False)
       .str.replace(",","",regex=False)
      ) #remove ₹ and , 
```

```python
price
```

```python
df["Email"].str.lower().str.contains("gmail") # search values
```

```python
df[df["Email"].str.lower().str.contains("gmail",na=False)] #shows only gmail values
```

```python
df["Course"] = df["Course_Code"].str.extract(r"([A-Z]+)")
df["Number"] = df["Course_Code"].str.extract(r"(\d+)")

df[["Course_Code", "Course", "Number"]] # extract part of Course_Code
```

```python
df["City"] = ["Ahm.edabad","sur.at","BARODA","mumBai"]
df["City"] = (
    df["City"]
      .astype("string")
      .str.strip()
      .str.lower()
      .str.replace(".", "", regex=False)
)
```
