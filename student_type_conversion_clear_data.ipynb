```python
import pandas as pd
```

```python
df=pd.read_csv("students_type_conversion.csv")
df1=df.copy()
```

```python
# text conversion
df1["StudentID"]=df["StudentID"].astype("string")
df1["Name"]=df["Name"].astype("string")
df1.info()
```

```python
# numeric conversion
df1["Age"]=pd.to_numeric(df1["Age"],errors="coerce").astype("Int64")
df1["Marks"]=pd.to_numeric(df1["Marks"],errors="coerce").astype("Float64")
df1["Fees"]=pd.to_numeric(df1["Fees"],errors="coerce").astype("Float64")
df1.info()
```

```python
# DateTime Conversion
df1["JoinDate"]=pd.to_datetime(
    df1["JoinDate"],format="mixed",errors="coerce"
)
df1["StudyTime"] = pd.to_timedelta(
    df1["StudyTime"],errors="coerce"
)
df1.info()
```

```python
# boolean conversion
active_map = {
    "yes":True,"y":True,"true":True,"1":True,
    "no":False,"n":False,"false":False,"0":False
}
df1["Active"] = (
    df1["Active"]
    .astype("string")
    .str.strip()
    .str.lower()
    .fillna("False")
    .map(active_map)
    .astype("boolean")
)
df1.info()
```

```python
# category conversion
df1["City"]=df1["City"].astype("category")
df1["City"].dtype
```
